```python
import pandas as pd
import numpy as np



#tabla en formato ancho raw como se estaba trabajando hasta la fecha
df = pd.read_excel('datosRAW_01.xlsx')
#indexamos la tabla
df = df.reset_index().rename(columns={'index': 'fila_original'})

df
```

```python
#nombres de las columnas de la base de datos, para copiar y pegar en los arrays
df.columns.to_list()
```

```python
# tabla de producción

#nombres de las columnas de produccion
cols_info = ['ID_produccion', 'Fecha', 'Maquina', 'Operario', 'Referencia', 'Turno']
#agrupamos las filas por id y sumamos los valores
df_produccion = df.groupby(cols_info, as_index=False)[['Pacas_producidas', 'Horas_trabajadas']].sum()

#ya que se presentan ID repetidos por errores humanos, vamos a tomar todos los ID repetidos y a sumar las columnas numericas
#y solo mantener la informacion de la primera fila que aparezca
df_produccion = (
    df_produccion.groupby('ID_produccion', as_index=False)
    .agg({
        'Fecha': 'first',
        'Maquina': 'first',
        'Turno': 'first',
        'Operario': 'first',
        'Referencia': 'first',
        'Pacas_producidas': 'sum',
        'Horas_trabajadas': 'sum'
    })
)

df_produccion.head()
```

```python
#extraccion unicamente de los nombres de las columnas que contengan codigo y en horas
cols_codigos = [c for c in df.columns if c.startswith("Codigo_") and c.endswith("_en_horas")]
cols_codigos
```

```python
#uso de melt para convertir las columnas en filas para generar una tabla, cada una es un posible paro por ID y numero de fila--> convertir en formato largo
df_paros = df.melt(
    id_vars=['fila_original', 'ID_produccion'],
    value_vars=[c for c in df.columns if c.startswith("Codigo_") and c.endswith("_en_horas")],
    var_name='Codigo_col',
    value_name='Tiempo_en_horas'
)
df_paros = df_paros.copy()
df_paros
```

```python
#crear una nueva columnas y extraer unicamente el numero en el calor de codigo de paro del codigo de paro
df_paros['Codigo_de_paro'] = df_paros['Codigo_col'].str.extract(r'Codigo_(\d+)_').astype(int)
df_paros
```

```python
#matris de booleanos para tomar unicamente las filas con dato en tiempo de los paros
df_paros = df_paros[df_paros['Tiempo_en_horas'] > 0]
df_paros
```

```python
#tabla auxiliar con la informacio de los subcodigos de paro y areas de codigo 5
# 2. Creamos tabla auxiliar con subcodigos y área
df_subinfo = df[['fila_original','Sub_Codigo_de_paro_1','Subcodigo_3','Subcodigo_5','Área_involucrada_en_subcodigo_5']]
```

```python
#merge de tabla paros con tabla auxiliar por el numero de fila
df_paros = df_paros.merge(df_subinfo, on='fila_original', how='left')
df_paros
```

```python
#insertar columnas vacias para luego rellenar con los datos de subcodigo y area
df_paros['Subcodigo'] = None
df_paros['Area_involucrada'] = None
df_paros
```

```python
#rellenamos los datos de subcodigo y merge pare mantener la integridad de los datos
df_paros.loc[df_paros['Codigo_de_paro'] == 1, 'Subcodigo'] = df_paros['Sub_Codigo_de_paro_1']
df_paros.loc[df_paros['Codigo_de_paro'] == 3, 'Subcodigo'] = df_paros['Subcodigo_3']
df_paros.loc[df_paros['Codigo_de_paro'] == 5, 'Subcodigo'] = df_paros['Subcodigo_5']

#rellenar los datos de area involucrada
df_paros.loc[df_paros['Codigo_de_paro'] == 5, 'Area_involucrada'] = df_paros['Área_involucrada_en_subcodigo_5']

df_paros
```

```python
#conservar unicamente las columnas de interes y organizar
df_paros = df_paros[['ID_produccion', 'Codigo_de_paro', 'Subcodigo', 'Area_involucrada', 'Tiempo_en_horas']]
df_paros
```

```python
#extrear ambas tablas en formato excel para su uso interno
df_produccion.to_excel("produccion_procesada.xlsx", index=False)
df_paros.to_excel("paros_procesados.xlsx", index=False)
```
