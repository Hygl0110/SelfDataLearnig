```python
import pandas as pd
import numpy as np



#tabla en formato ancho raw como se estaba trabajando hasta la fecha
df = pd.read_excel('datosRAW_01.xlsx')
df
```

```python
#nombres de las columnas de la base de datos, para copiar y pegar en los arrays
df.columns.to_list()
```

```python
#tabla de porduccion
#nombres de las columnas de produccion
cols_info = ['ID_produccion', 'Fecha', 'Maquina', 'Operario', 'Referencia', 'Turno']

df_produccion = df.groupby(cols_info, as_index=False)[['Pacas_producidas', 'Horas_trabajadas']].sum()
df_produccion.head()
```

```python
#extraccion unicamente de los nombres de las columnas que contengan codigo y en horas
cols_paros = [c for c in df.columns if c.startswith("Codigo_") and c.endswith("_en_horas")]
cols_paros
```

```python
#uso de melt para convertir las columnas en filas para generar una tabla, cada una es un posible paro
df_paros = df.melt(
    id_vars=['ID_produccion'],
    value_vars=cols_paros,
    var_name='Codigo_columna',
    value_name='Tiempo_en_horas'
)
df_paros.head()
```

```python
#crear una nueva columnas y extrar sunicamente el numero del codigo de paro
df_paros['Codigo_de_paro'] = df_paros['Codigo_columna'].str.extract(r'(\d+)').astype(int)
df_paros.head()
```

```python
df_paros = df_paros[df_paros['Tiempo_en_horas'] > 0]
df_paros
```

```python
#insertar los subcodigos a la tabla de paros iniciando todos con NA
df_paros['Subcodigo'] = "NA"
df_paros.head()
```

```python
#relacionar el valor de subcodigo de la tabla df, con cada codigo para cada codigo 1, 3 y 5
df_paros.loc[df_paros['Codigo_de_paro'] == 1, 'Subcodigo'] = df['Sub_Codigo_de_paro_1']
df_paros.loc[df_paros['Codigo_de_paro'] == 3, 'Subcodigo'] = df['Subcodigo_3']
df_paros.loc[df_paros['Codigo_de_paro'] == 5, 'Subcodigo'] = df['Subcodigo_5']
df_paros
```

```python
#conservar unicamente las columnas de interes y organizar
df_paros = df_paros[['ID_produccion', 'Codigo_de_paro', 'Subcodigo', 'Tiempo_en_horas']]
df_paros
```

```python
df_produccion.to_excel("produccion_procesada.xlsx", index=False)
df_paros.to_excel("paros_procesados.xlsx", index=False)
```
